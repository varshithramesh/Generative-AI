```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch
```

## Loading the GPT-2 Model

```python
model_name = "gpt2"

tokenizer = GPT2Tokenizer.from_pretrained(model_name)
model = GPT2LMHeadModel.from_pretrained(model_name)

model.eval()

print("GPT-2 model loaded successfully!")
```

##  Creating a Text Generation Function

A text generation function is created so that different generation parameters can be easily changed for each experiment. The function uses temperature, top-k, top-p, and repetition penalty to control the generated text.

```python
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0,
                  repetition_penalty=1.0, max_length=100):

    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_length=max_length,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=repetition_penalty,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)

    return generated_text
```

## Original Prompt

**Original Prompt:**

"The young scientist entered the abandoned laboratory and discovered"

The same prompt is used for all five experiments so that the effect of changing the generation parameters can be compared.

## Experiment 1 – Temperature

**Parameter Changed:** Temperature

**Value Used:** 0.5

**Other Parameters:** Kept at their default values.

```python
prompt = "The young scientist entered the abandoned laboratory and discovered"

output1 = generate_text(
    prompt,
    temperature=0.5,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print(output1)
```



### Observation

The lower temperature produced a more focused and predictable continuation. The output stayed closer to the given laboratory setting, although some repetition may be present.

##  Experiment 2 – Top-k

**Parameter Changed:** Top-k

**Value Used:** 10

**Other Parameters:** Temperature = 1.0, Top-p = 1.0, Repetition Penalty = 1.0

```python
output2 = generate_text(
    prompt,
    temperature=1.0,
    top_k=10,
    top_p=1.0,
    repetition_penalty=1.0
)

print(output2)
```



### Observation

The output was more controlled because top-k limited the number of possible words considered for the next token. The continuation remained related to the original laboratory setting, although some unexpected ideas could appear.

##  Experiment 3 – Top-p

**Parameter Changed:** Top-p

**Value Used:** 0.7

**Other Parameters:** Temperature = 1.0, Top-k = 50, Repetition Penalty = 1.0

```python
output3 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=0.7,
    repetition_penalty=1.0
)

print(output3)
```



### Observation

The output was descriptive and created a fairly clear story-like continuation. The top-p setting allowed the model to choose from a controlled group of likely words while still providing some creativity.

##  Experiment 4 – Repetition Penalty

**Parameter Changed:** Repetition Penalty

**Value Used:** 1.5

**Other Parameters:** Temperature = 1.0, Top-k = 50, Top-p = 1.0

```python
output4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.5
)

print(output4)
```



### Observation

The repetition penalty encouraged the model to avoid repeating the same words and phrases. The output used more varied wording, although reducing repetition did not always guarantee a completely coherent continuation.

##  Experiment 5 – High Temperature

**Parameter Changed:** Temperature

**Value Used:** 1.5

This experiment uses a high temperature to observe whether increasing randomness produces a more creative or unusual output.

```python
output5 = generate_text(
    prompt,
    temperature=1.5,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print(output5)
```



### Observation

The high temperature produced a more random and unusual continuation. The output was more creative, but it could become difficult to follow because the model had greater freedom in selecting the next words.
